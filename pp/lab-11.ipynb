### 05) WAP to generate 3 random integers between 100 and 999 which is divisible by 5.

```python
import random as r

gen = lambda: r.randrange(100, 999, 5)

a = gen()
b = gen()
c = gen()

assert a % 5 == 0, "A isn't divisible by 5"
assert b % 5 == 0, "B isn't divisible by 5"
assert c % 5 == 0, "C isn't divisible by 5"
print(a, b, c)
```

### 06) WAP to generate 100 random lottery tickets and pick two lucky tickets from it and announce them as Winner and Runner up respectively.

```python
import random as r

tickets = [_ for _ in range(1, 101)]

first_winner, runner_up = r.sample(tickets, k=2)

print(first_winner)
print(runner_up)
```

### 08) Subtract a week (7 days) from a given date in Python.

```python
from datetime import datetime, timedelta

now = datetime.now()

print(now - timedelta(days=7))
```

### 09) WAP to Calculate number of days between two given dates.

```python
from datetime import datetime, timedelta

now = datetime.now()
ndate = now - timedelta(days=7, minutes=30)

print((now - ndate).days)
```

### 10) WAP to Find the day of the week of a given date.(i.e. wether it is sunday/monday/tuesday/etc.)

```python
from datetime import datetime

now = datetime.now()
days = ['MON', 'TUE', 'WENS', 'THUS', 'FRI', 'SAT', 'SUN']

print(days[now.weekday()])
```
