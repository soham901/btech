### 09) WAP to find the distance between two points in 2-D space.

```python
import math

x1 = int(input("for first point\nEnter x : "))
y1 = int(input("Enter y : "))

x2 = int(input("\n\nfor second point\nEnter x : "))
y2 = int(input("Enter y : "))

print(math.sqrt(math.pow((x1-x2), 2) + math.pow((y1-y2), 2)))
```

### 10) WAP to print sum of n natural numbers.

### 11) WAP to print sum of square of n natural numbers.

```python
import math

n = int(input("Enter n : "))
sum = 0

for a in range(1, n+1):
    sum += math.pow(a, 2)

print(sum)
```

### 14) WAP to get the distance from user into kilometer, and convert it into meter, feet, inches and centimeter.

```python
km = float(input("Enter KM : "))

meters = km * 1000
feet = km * 3280.84
inches = km * 39370.1
centimeters = km * 100000

print(f"Distance in meters: {meters}")
print(f"Distance in feet: {feet}")
print(f"Distance in inches: {inches}")
print(f"Distance in centimeters: {centimeters}")
```
