### 01) Write a function to calculate BMI given mass and height. (BMI = mass/h**2)

```python
def calc_bmi(height, weight):
    res = weight / (height * height)
    return res

calc_bmi(1.8, 80)
```

### 02) Write a function that add first n numbers.

```python
def sum_of_first(n):
    return (n * (n + 1)) / 2

sum_of_first(3)
```

### 03) Write a function that returns 1 if the given number is Prime or 0 otherwise.

```python
def is_prime(n):
    for i in range(2, n):
        if n % i == 0: return 0
    return 1

is_prime(4)
```

### 04) Write a function that returns the list of Prime numbers between given two numbers.

```python
def prime_between(a, b):
    res = []
    for i in range(a, b + 1):
        if is_prime(i):
            res.append(i)
    return res

prime_between(1, 10)
```

### 05) Write a function that returns True if the given string is Palindrome or False otherwise.

```python
def is_palidrome(s):
    return s == s[::-1]

print(
    is_palidrome("hello"),
    is_palidrome("madam")
)
```

### 06) Write a function that returns the sum of all the elements of the list.

```python
def sum_of_list(l):
    res = 0
    for i in l:
        res += i
    return res

sum_of_list([1, 2, 3])
```

### 07) Write a function to calculate the sum of the first element of each tuples inside the list.

```python
def sum_of_tuple(lst):
    res = [a[0] for a in lst]
    return sum_of_list(res)

sum_of_tuple([(1, 2), (2, 3)])
```

### 08) Write a recursive function to find nth term of Fibonacci Series.

```python
def fibo(n):
    if n in [0, 1]: return n
    return fibo(n - 1) + fibo(n - 2)

fibo(1)
```

### 10) Write a function to get the sum of the scores ending with zero.
#### Example : scores = [200, 456, 300, 100, 234, 678] 
#### Ans = 200 + 300 + 100 = 600

```python
scores = [200, 456, 300, 100, 234, 678]

def sum_of_scores(scores):
    total = 0
    for a in scores:
        total += a
    return total

sum_of_scores(scores)
```

### 11) Write a function to invert a given Dictionary.
#### hint: keys to values & values to keys
#### Before : {'a': 10, 'b':20, 'c':30, 'd':40}
#### After : {10:'a', 20:'b', 30:'c', 40:'d'}

```python
og_dict = {'a': 10, 'b':20, 'c':30, 'd':40}

def transform_dict(_dict):
    return {v:k for k, v in _dict.items()}

transform_dict(og_dict)
```

### 12) Write a function to check whether the given string is Pangram or not.
#### hint: Pangram is a string containing all the characters a-z atlest once.
#### "the quick brown fox jumps over the lazy dog" is a Pangram string.

```python
def is_panagram(s):
    return set("abcdefghijklmnopqrstuvwxyz").issubset(set(s.lower()))

is_panagram("The quick brown fox jumps over the lazy dog")
```

### 13) Write a function that returns the number of uppercase and lowercase letters in the given string. 
#### example : Input : s1 = AbcDEfgh ,Ouptput : no_upper = 3, no_lower = 5

```python
def count_cases(s):
    upper = 0
    lower = 0
    for i in s:
        if i.isupper(): upper += 1
        if i.islower(): lower += 1
    return upper, lower

count_cases("The quick brown fox jumps over the lazy dog")
```

### 14) Write a lambda function to get smallest number from the given two numbers.

```python
smallest_num = lambda a, b: a if a < b else b

smallest_num(1, -2)
```

### 15) For the given list of names of students, extract the names having more that 7 characters. Use filter().

```python
def long_names(names):
    return list(filter(lambda a: len(a) > 7, names))

long_names(["Abhishek", "Soham"])
```

### 16) For the given list of names of students, convert the first letter of all the names into uppercase. use map().

```python
def title_case_list(names):
    return list(map(lambda a: a.title(), names))

title_case_list(["abhishek", "soham"])
```
