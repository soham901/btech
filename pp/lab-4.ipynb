# String

### 01) WAP to check whether the given string is palindrome or not.

```python
str = input('Enter: ')
print('YES' if str == str[::-1] else 'NO')
```

### 02) WAP to reverse the words in the given string.

```python
str = input('Enter: ')

print(' '.join(str.split()[::-1]))
```

### 03) WAP to remove ith character from given string.

```python
str = input('Enter: ')
ch = int(input('Enter index: '))
print(str[:ch] + str[ch+1:])
```

### 04) WAP to find length of string without using len function.

```python
str = input('Enter: ')

c = 0
for ch in str:
    c += 1

print(c)
```

### 05) WAP to print even length word in string.

```python
str = input('Enter: ').split()
for ch in str:
    if len(ch) % 2 == 0:
        print(ch)
```

### 06) WAP to count numbers of vowels in given string.

```python
str = input('Enter: ')
VOWELS = ['a', 'e', 'i', 'o', 'u']
c = 0
for ch in str:
    if ch in VOWELS:
        c += 1

print(c)
```

### 07) WAP to capitalize the first and last character of each word in a string.

```python
str = input('Enter: ').split()

for ch in str:
    print(ch[0].upper() + ch[1:-1] + ch[-1].upper())
```

### 08) WAP to convert given array to string.

```python
arr = [10, 20, 30, 40, 50]

print(' '.join(map(str, arr)))
```

### 09) Check if the password and confirm password is same or not. 
### In case of only case's mistake, show the error message.

```python
password = input('Enter password: ').strip()
cpassword = input('Enter confirm password: ').strip()
flag = True

if len(password) < 6:
    print('Password must be at least 6 characters long')
    flag = False

if password != cpassword:
    print('Passwords do not match')
    flag = False

if flag:
    print('Password is valid')
```

### 10) : Display credit card number. 
### card no. : 1234 5678 9012 3456
### display as  : **** **** **** 3456

```python
cno = input('Enter the card number : ').split(' ')
result = ''

for ch in range(len(cno)-1):
    ch = cno[ch]
    result += ch.replace(ch[::], '*'*len(ch)+' ')

result += cno[-1]
print(result)
```

### 11) : Checking if the two strings are Anagram or not.
### s1 = decimal and s2 = medical are Anagram

```python
s1 = input('Enter a str 1: ')
s2 = input('Enter a str 2: ')

print(f'{s1} and {s2} are {"" if sorted(s1) == sorted(s2) else "not "}Anagram')
```

### 12) : Rearrange the given string. First lowercase then uppercase alphabets.
### input : EHlsarwiwhtwMV
### output : lsarwiwhtwEHMV

```python
str = input('Enter the str: ')
up = ''
low = ''

for ch in str:
    if ch.isupper(): up += ch
    else: low += ch

print(low + up)
```
