### 01) WAP to read and display the contents of a text file. (also try to open the file in some other directory) 
#### - in the form of a string
#### - line by line
#### - in the form of a list

```python
def custom_read(_fp, mode, line_no=0):
    match mode:
        case "plain":
            return fp.read()
        case "lines":
            fp.seek(0)
            return fp.readlines()[line_no]
        case "list":
            fp.seek(0)
            return fp.readlines()
        case _:
            raise ValueError(f"Invalid mode: {mode}")


fp = open("docker-compose.yml", "r")

print(custom_read(fp, "plain"))
print(custom_read(fp, "lines"))
print(custom_read(fp, "list"))
```

```python
LOG_FILE = "./demo.log"

def persist_log(*args, **kwargs):
    with open(LOG_FILE, "a") as f:
        f.write(*args, **kwargs)
        f.write("\n")

def log(*args, **kwargs):
    print(*args, **kwargs)
    persist_log(*args, **kwargs)


log("Hello")
```

### 04) WAP to find the longest word(s) in a file

```python
def longest_word_in_file(fp):
    fp.seek(0)
    return max(fp.read().split(), key=len)

fp = open("docker-compose.yml", "r")
print(longest_word_in_file(fp))
```

### 05) WAP to count the no. of lines, words and characters in a given text file.

```python
def stats(fp):
    fp.seek(0)
    lines = fp.readlines()
    fp.seek(0)
    words = fp.read().split()
    fp.seek(0)
    chars = fp.read()
    return {
        "lines": len(lines),
        "words": len(words),
        "chars": len(chars)
    }

fp = open("docker-compose.yml", "r")
print(stats(fp))
```

### 06) WAP to copy the content of a file to the another file.

```python
def backup_log(fp):
    fp.seek(0)
    with open(f'backup-{fp.name}', "w") as f:
        f.write(fp.read())

fp = open("docker-compose.yml", "r")
backup_log(fp)
```

### 08) WAP to create an UDF named frequency to count occurances of the specific word in a given text file.

```python
def freq_of_words(fp, word):
    fp.seek(0)
    words = fp.read().split()
    return words.count(word)

fp = open("docker-compose.yml", "r")
print(freq_of_words(fp, "jupyter"))
```

### 09) WAP to get the score of five subjects from the user, store them in a file. Fetch those marks and find the highest score.

```python
class ResultManager:
    def __init__(self):
        self.__marks = self.load_marks()
        
    def get_marks(self):
        return self.__marks
    
    def set_marks(self, subject, marks):
        self.__marks[subject] = marks
    
    def save_marks(self):
        with open("result.txt", "w") as f:
            f.write(str(self.__marks))
        
    def load_marks(self):
        if not os.path.exists("result.txt"):
            return {}
        with open("result.txt", "r") as f:
            return eval(f.read())
    
    def get_highest_score(self):
        return max(self.__marks.values())
    
    def flush(self):
        self.__marks = {}
        self.save_marks()


result_manager = ResultManager()

# ask the user whether to flush the result or just load it
choice = input("Do you want to flush the result? (y/n): ")

if choice.lower() == "y":
    result_manager.flush()
    for i in range(5):
        subject = input("Enter subject: ")
        marks = int(input("Enter marks: "))
        result_manager.set_marks(subject, marks)
    result_manager.save_marks()
    
print(result_manager.get_marks())
print("Highest score: ", result_manager.get_highest_score())
```

### 10) WAP to write first 100 prime numbers to a file named primenumbers.txt 
(Note: each number should be in new line)

```python
def first_n_prime_numbers(n):
    primes = []
    num = 2
    while len(primes) < n:
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
        num += 1
    return primes


fp = open("first_100_primes.txt", "w")
fp.write(str(first_n_prime_numbers(100)))
fp.close()
```

### 11) WAP to merge two files and write it in a new file.

```python
def merge_two_files(fp1, fp2, sep):
    fp1.seek(0)
    fp2.seek(0)
    return fp1.read() + sep + fp2.read()

fp1 = open("first_100_primes.txt", "r")
fp2 = open("first_100_primes.txt", "r")
fp3 = open("first_200_primes.txt", "w")
fp3.write(merge_two_files(fp1, fp2, "\n\n"))
fp3.close()
```

### 12) WAP to replace word1 by word2 of a text file. Write the updated data to new file.

```python
def replace_word_in_file(fp, old_word, new_word):
    fp.seek(0)
    lines = fp.readlines()
    fp.seek(0)
    with open(fp.name, "w") as f:
        for line in lines:
            f.write(line.replace(old_word, new_word))

fp = open("first_100_primes.txt", "r")
replace_word_in_file(fp, "2", "two")
```
